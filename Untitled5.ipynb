```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('lab5.png')

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
sigma = 2
amount = 1.5
```

```python
image_float = image.astype(np.float32) / 255.0
```

```python
blurred = cv2.GaussianBlur(
    image_float,
    (0, 0),
    sigmaX=sigma,
    sigmaY=sigma
)
```

```python
sharpened = np.clip(
    image_float + amount * (image_float - blurred),
    0,
    1
)
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(blurred)
ax[1].set_title("Blurred")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened")
ax[2].axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt


image = cv2.imread('lab5.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Gaussian 5x5
gaussian1 = cv2.GaussianBlur(image, (5,5), 0)

# Gaussian 21x21
gaussian2 = cv2.GaussianBlur(image, (21,21), 0)


fig, ax = plt.subplots(1,3, figsize=(15,5))

ax[0].imshow(image)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(gaussian1)
ax[1].set_title("Gaussian 5x5")
ax[1].axis("off")

ax[2].imshow(gaussian2)
ax[2].set_title("Gaussian 21x21")
ax[2].axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('lab5.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

image_float = image.astype(np.float32) / 255.0
```

```python
laplacian = cv2.Laplacian(image_float, cv2.CV_32F)
```

```python
sharpened = np.clip(image_float - laplacian, 0, 1)
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))

ax[0].imshow(image_float)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(laplacian)
ax[1].set_title("Laplacian")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened")
ax[2].axis("off")

plt.show()
```
